#### Program 9: Comparing Pauli-X, Pauli-Y and Pauli-Z gates

```python
%matplotlib inline
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector
from qiskit.visualization import plot_bloch_multivector

# Use a 3-qubit register: qubit0 -> X, qubit1 -> Y, qubit2 -> Z
# so all three Bloch spheres can be compared side-by-side in one figure.
qc = QuantumCircuit(3)
qc.x(0)
qc.y(1)
qc.z(2)
print(qc.draw())

state = Statevector.from_instruction(qc)
print("Statevector:", state)

fig = plot_bloch_multivector(
    state, title="Program 9: q0=X  q1=Y  q2=Z (compared)"
)
fig.savefig("/home/claude/ws2/images/p9_compare_xyz.png", bbox_inches="tight")
fig
```
