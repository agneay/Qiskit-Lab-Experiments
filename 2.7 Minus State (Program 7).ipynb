#### Program 7: |−⟩ state using Hadamard and Pauli-Z gates

```python
%matplotlib inline
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector
from qiskit.visualization import plot_bloch_multivector

qc = QuantumCircuit(1)
qc.h(0)
qc.z(0)                       # |0> --H--> |+> --Z--> |-> = (|0>-|1>)/sqrt(2)
print(qc.draw())

state = Statevector.from_instruction(qc)
print("Statevector:", state)

fig = plot_bloch_multivector(state, title="Program 7: |-> state")
fig.savefig("/home/claude/ws2/images/p7_minus_state.png", bbox_inches="tight")
fig
```
