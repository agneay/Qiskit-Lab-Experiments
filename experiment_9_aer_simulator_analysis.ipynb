### PROCEDURE / CODE

**Import required libraries**

```python
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator
from qiskit.visualization import plot_histogram
```

**Construct a Bell state circuit to simulate**

```python
qc = QuantumCircuit(2, 2)
qc.h(0)
qc.cx(0, 1)
qc.measure([0, 1], [0, 1])

print(qc.draw())
```

**Run the simulation on the Aer Simulator**

```python
simulator = AerSimulator()
job = simulator.run(qc, shots=2048)
result = job.result()
counts = result.get_counts()

print("Measurement counts:", counts)
```

**Analyze the results**

```python
plot_histogram(counts)
```

**Analysis:** The measurement outcomes are concentrated on the `00` and `11` states with roughly equal probability, confirming the entangled Bell state produced by the Hadamard and CNOT gates. This demonstrates the correlated, probabilistic nature of quantum measurement as simulated by the Aer Simulator.
