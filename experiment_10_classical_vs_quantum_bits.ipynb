### PROCEDURE / CODE

**Import required libraries**

```python
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator
from qiskit.visualization import plot_histogram
```

**Classical-like behavior: qubit without superposition**

```python
# A qubit that is never put into superposition behaves like a classical bit.
# It always yields a deterministic (single) outcome upon measurement.
qc_classical = QuantumCircuit(1, 1)
qc_classical.measure(0, 0)   # qubit stays in |0>, deterministic result

simulator = AerSimulator()
counts_classical = simulator.run(qc_classical, shots=1024).result().get_counts()
print("Classical-like bit measurement counts:", counts_classical)
```

**Quantum behavior: qubit in superposition**

```python
# Applying a Hadamard gate creates superposition, giving a probabilistic outcome.
qc_quantum = QuantumCircuit(1, 1)
qc_quantum.h(0)
qc_quantum.measure(0, 0)

counts_quantum = simulator.run(qc_quantum, shots=1024).result().get_counts()
print("Quantum bit (qubit) measurement counts:", counts_quantum)
```

**Compare the two results side by side**

```python
plot_histogram([counts_classical, counts_quantum], legend=['Classical-like bit', 'Quantum bit'])
```

**Observation:** The classical-like bit always measures `0` with 100% probability (deterministic), whereas the quantum bit in superposition measures `0` and `1` with roughly equal (~50%) probability each run, illustrating the fundamental difference between classical bits and qubits.
