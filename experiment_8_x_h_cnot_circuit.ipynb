### PROCEDURE / CODE

**Import required libraries**

```python
from qiskit import QuantumCircuit
```

**Design a circuit using X, H, and CNOT gates**

```python
# Create a quantum circuit with 2 qubits
qc = QuantumCircuit(2)

# Apply Pauli-X gate on qubit 0
qc.x(0)

# Apply Hadamard gate on qubit 1
qc.h(1)

# Apply CNOT gate: qubit 1 (control) -> qubit 0 (target)
qc.cx(1, 0)

# Display the circuit diagram
print(qc.draw())
```

**Visualize the circuit graphically**

```python
qc.draw('mpl')
```
