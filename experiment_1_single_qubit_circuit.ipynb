### PROCEDURE / CODE

**Import required libraries**

```python
from qiskit import QuantumCircuit
```

**Create a single-qubit quantum circuit**

```python
# Create a quantum circuit with 1 qubit
qc = QuantumCircuit(1)

# Display the circuit diagram (text form)
print(qc.draw())
```

**Display the circuit diagram graphically**

```python
qc.draw('mpl')
```
