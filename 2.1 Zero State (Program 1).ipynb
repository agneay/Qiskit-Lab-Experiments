#### Program 1: Single qubit in the |0⟩ state visualized on the Bloch Sphere

```python
%matplotlib inline
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector
from qiskit.visualization import plot_bloch_multivector

# Single-qubit circuit, qubit initialised by default in |0>
qc = QuantumCircuit(1)
print(qc.draw())

state = Statevector.from_instruction(qc)
print("Statevector:", state)

fig = plot_bloch_multivector(state, title="Program 1: |0> state")
fig.savefig("/home/claude/ws2/images/p1_zero_state.png", bbox_inches="tight")
fig
```
